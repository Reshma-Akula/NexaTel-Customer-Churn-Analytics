Step 4 — Automatically check primary-key uniqueness and foreign-key validity using Python.(Validate the 22 straightforward key candidates)

```python
# Step 4A — Primary Key Uniqueness Check

primary_keys = {
    "billing": "invoice_id",
    "cities": "city_id",
    "complaints": "complaint_id",
    "contracts": "contract_id",
    "customers": "customer_id",
    "customer_feedback": "feedback_id",
    "devices": "device_id",
    "employees": "employee_id",
    "marketing_campaigns": "campaign_id",
    "network_quality": "record_id",
    "payments": "payment_id",
    "plans": "plan_id",
    "plan_history": "history_id",
    "recharges": "recharge_id",
    "regions": "region_id",
    "states": "state_id",
    "stores": "store_id",
    "subscriptions": "subscription_id",
    "support_tickets": "ticket_id",
    "usage_data": "usage_id",
    "usage_sms": "usage_id",
    "usage_voice": "usage_id"
}

print("PRIMARY KEY UNIQUENESS CHECK")
print("=" * 80)

for table, key in primary_keys.items():
    df = tables[table]

    total_rows = len(df)
    unique_values = df[key].nunique()
    null_values = df[key].isna().sum()
    duplicate_values = df[key].duplicated().sum()

    print(
        f"{table:25} | "
        f"Rows: {total_rows:7} | "
        f"Unique: {unique_values:7} | "
        f"Nulls: {null_values:5} | "
        f"Duplicates: {duplicate_values:5}"

    )
```

```python
# Step 4A.1 — Investigate duplicate customer IDs

customers_df = tables["customers"]

duplicate_customers = customers_df[
    customers_df["customer_id"].duplicated(keep=False)
].sort_values("customer_id")

print("Duplicate customer records:")
print(duplicate_customers[["customer_id", "first_name", "last_name",
                           "plan_id", "customer_status", "acquisition_date",
                           "churn_date"]].to_string(index=False))

print("\nNumber of duplicate customer IDs:",
      duplicate_customers["customer_id"].nunique())
```

Step 4B — Foreign Key Validation

```python
# Step 4B —  Foreign Key Validation

foreign_keys = [
    ("billing", "customer_id", "customers", "customer_id"),
    ("complaints", "customer_id", "customers", "customer_id"),
    ("contracts", "customer_id", "customers", "customer_id"),
    ("customer_feedback", "customer_id", "customers", "customer_id"),
    ("devices", "customer_id", "customers", "customer_id"),
    ("network_quality", "customer_id", "customers", "customer_id"),
    ("payments", "customer_id", "customers", "customer_id"),
    ("payments", "invoice_id", "billing", "invoice_id"),
    ("plan_history", "customer_id", "customers", "customer_id"),
    ("recharges", "customer_id", "customers", "customer_id"),
    ("subscriptions", "customer_id", "customers", "customer_id"),
    ("support_tickets", "customer_id", "customers", "customer_id"),
    ("usage_data", "customer_id", "customers", "customer_id"),
    ("usage_sms", "customer_id", "customers", "customer_id"),
    ("usage_voice", "customer_id", "customers", "customer_id"),

    ("cities", "state_id", "states", "state_id"),
    ("cities", "region_id", "regions", "region_id"),

    ("stores", "city_id", "cities", "city_id"),
    ("stores", "state_id", "states", "state_id"),
    ("stores", "region_id", "regions", "region_id"),

    ("employees", "store_id", "stores", "store_id"),
    ("employees", "region_id", "regions", "region_id"),

    ("support_tickets", "assigned_employee_id", "employees", "employee_id"),

    ("customers", "city_id", "cities", "city_id"),
    ("customers", "region_id", "regions", "region_id"),
    ("customers", "plan_id", "plans", "plan_id"),

    ("contracts", "plan_id", "plans", "plan_id"),
    ("subscriptions", "plan_id", "plans", "plan_id"),
    ("recharges", "plan_id", "plans", "plan_id")
]

print("FOREIGN KEY VALIDATION")
print("=" * 100)

for child_table, child_col, parent_table, parent_col in foreign_keys:

    child_df = tables[child_table]
    parent_df = tables[parent_table]

    parent_values = set(
        parent_df[parent_col].dropna().astype(str)
    )

    child_values = child_df[child_col].dropna().astype(str)

    invalid_count = (~child_values.isin(parent_values)).sum()
    null_count = child_df[child_col].isna().sum()

    print(
        f"{child_table:22}."
        f"{child_col:25} -> "
        f"{parent_table:22}."
        f"{parent_col:15} | "
        f"Invalid: {invalid_count:5} | "
        f"Nulls: {null_count:5}"
    )
```

```python
# Step 4C — Validate Remaining Keys and Relationships

print("REMAINING KEY & RELATIONSHIP VALIDATION")
print("=" * 90)

# 1. Plan history → Plans
print("\n1. PLAN HISTORY → PLANS")

plans = set(tables["plans"]["plan_id"].dropna().astype(str))

for column in ["old_plan_id", "new_plan_id"]:
    values = tables["plan_history"][column].dropna().astype(str)
    invalid = (~values.isin(plans)).sum()
    nulls = tables["plan_history"][column].isna().sum()

    print(
        f"plan_history.{column:12} -> plans.plan_id"
        f" | Invalid: {invalid:5} | Nulls: {nulls:5}"
    )


# 2. Retention campaigns → Customers
print("\n2. RETENTION CAMPAIGNS → CUSTOMERS")

customers = set(tables["customers"]["customer_id"].dropna().astype(str))

retention_customers = tables["retention_campaigns"]["customer_id"].dropna().astype(str)

invalid = (~retention_customers.isin(customers)).sum()
nulls = tables["retention_campaigns"]["customer_id"].isna().sum()

print(
    f"retention_campaigns.customer_id -> customers.customer_id"
    f" | Invalid: {invalid:5} | Nulls: {nulls:5}"
)


# 3. Retention campaign IDs
print("\n3. RETENTION CAMPAIGN ID CHECK")

retention = tables["retention_campaigns"]

print("Rows:", len(retention))
print("Unique campaign_id:", retention["campaign_id"].nunique())
print("Duplicate campaign_id values:", retention["campaign_id"].duplicated().sum())

# Check whether campaign_id + customer_id is unique
composite_duplicates = retention.duplicated(
    subset=["campaign_id", "customer_id"]
).sum()

print(
    "Duplicate campaign_id + customer_id combinations:",
    composite_duplicates
)


# 4. Billing invoice IDs
print("\n4. BILLING INVOICE ID CHECK")

billing = tables["billing"]

print("Rows:", len(billing))
print("Unique invoice_id:", billing["invoice_id"].nunique())
print("Duplicate invoice_id values:", billing["invoice_id"].duplicated().sum())


# 5. Payments payment IDs
print("\n5. PAYMENT ID CHECK")

payments = tables["payments"]

print("Rows:", len(payments))
print("Unique payment_id:", payments["payment_id"].nunique())
print("Duplicate payment_id values:", payments["payment_id"].duplicated().sum())


# 6. Check whether duplicate billing rows are exact duplicates
print("\n6. EXACT DUPLICATE BILLING ROWS")

billing_exact_duplicates = billing.duplicated().sum()

print("Exact duplicate billing rows:", billing_exact_duplicates)


# 7. Check whether duplicate payment rows are exact duplicates
print("\n7. EXACT DUPLICATE PAYMENT ROWS")

payment_exact_duplicates = payments.duplicated().sum()

print("Exact duplicate payment rows:", payment_exact_duplicates)
```

# Step 7C — Customer Data Cleaning

The customer data will be cleaned using the findings from the investigation.

Cleaning principles:

1. Remove exact duplicate customer records.
2. Standardize email and phone fields.
3. Flag invalid email and phone values rather than inventing replacement values.
4. Standardize PIN codes and flag invalid PINs.
5. Recalculate age from date of birth where possible.
6. Standardize mixed date formats.
7. Preserve legitimate missing values such as churn_date for active customers.
8. Preserve the original raw customer table unchanged.

The resulting table will be used as the analytical customer dimension.

```python
# Step 7C — Clean Customer Dimension

customers_clean = tables["customers"].copy()

print("CUSTOMER CLEANING")
print("=" * 80)

# --------------------------------------------------
# 1. Remove exact duplicate customer records
# --------------------------------------------------

before = len(customers_clean)

customers_clean = customers_clean.drop_duplicates()

after = len(customers_clean)

print("Exact duplicate rows removed:", before - after)


# --------------------------------------------------
# 2. Standardize text fields
# --------------------------------------------------

text_columns = [
    "first_name",
    "last_name",
    "gender",
    "occupation",
    "customer_segment",
    "product_line",
    "acquisition_channel",
    "customer_status"
]

for col in text_columns:
    customers_clean[col] = (
        customers_clean[col]
        .astype("string")
        .str.strip()
        .str.title()
    )


# --------------------------------------------------
# 3. Clean email
# --------------------------------------------------

customers_clean["email"] = (
    customers_clean["email"]
    .astype("string")
    .str.strip()
    .str.lower()
)

email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

invalid_email_mask = (
    customers_clean["email"].notna()
    & ~customers_clean["email"].str.match(
        email_pattern,
        na=False
    )
)

customers_clean.loc[
    invalid_email_mask,
    "email"
] = pd.NA

print("Invalid emails converted to null:", invalid_email_mask.sum())


# --------------------------------------------------
# 4. Clean phone numbers
# --------------------------------------------------

customers_clean["phone_number"] = (
    customers_clean["phone_number"]
    .astype("string")
    .str.strip()
)

phone_digits = customers_clean["phone_number"].str.replace(
    r"\D",
    "",
    regex=True
)

valid_phone_mask = phone_digits.str.len().eq(12)

invalid_phone_mask = (
    customers_clean["phone_number"].notna()
    & ~valid_phone_mask
)

customers_clean.loc[
    invalid_phone_mask,
    "phone_number"
] = pd.NA

print(
    "Invalid phone numbers converted to null:",
    invalid_phone_mask.sum()
)


# --------------------------------------------------
# 5. Clean PIN code
# --------------------------------------------------

pin_digits = (
    customers_clean["pin_code"]
    .astype("string")
    .str.replace(r"\D", "", regex=True)
)

valid_pin_mask = pin_digits.str.len().eq(6)

invalid_pin_mask = (
    customers_clean["pin_code"].notna()
    & ~valid_pin_mask
)

customers_clean.loc[
    invalid_pin_mask,
    "pin_code"
] = pd.NA

customers_clean["pin_code"] = pin_digits.where(
    valid_pin_mask,
    pd.NA
)

print("Invalid PINs converted to null:", invalid_pin_mask.sum())


# --------------------------------------------------
# 6. Standardize dates
# --------------------------------------------------

customers_clean["acquisition_date"] = pd.to_datetime(
    customers_clean["acquisition_date"],
    format="mixed",
    errors="coerce"
)

customers_clean["churn_date"] = pd.to_datetime(
    customers_clean["churn_date"],
    format="mixed",
    errors="coerce"
)

customers_clean["date_of_birth"] = pd.to_datetime(
    customers_clean["date_of_birth"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)


# --------------------------------------------------
# 7. Recalculate invalid ages from date of birth
# --------------------------------------------------

reference_date = pd.Timestamp("2026-08-24")

dob = customers_clean["date_of_birth"]

calculated_age = (
    reference_date.year
    - dob.dt.year
    - (
        (
            dob.dt.month > reference_date.month
        )
        |
        (
            (dob.dt.month == reference_date.month)
            &
            (dob.dt.day > reference_date.day)
        )
    ).astype(int)
)

invalid_age_mask = (
    (customers_clean["age"] < 18)
    | (customers_clean["age"] > 100)
)

customers_clean.loc[
    invalid_age_mask,
    "age"
] = calculated_age[invalid_age_mask]

print(
    "Invalid ages corrected:",
    invalid_age_mask.sum()
)


# --------------------------------------------------
# 8. Final validation
# --------------------------------------------------

print("\nFINAL CUSTOMER VALIDATION")
print("=" * 80)

print("Rows:", len(customers_clean))

print(
    "Unique customer IDs:",
    customers_clean["customer_id"].nunique()
)

print(
    "Duplicate customer IDs:",
    customers_clean["customer_id"].duplicated().sum()
)

remaining_invalid_age = (
    (customers_clean["age"] < 18)
    | (customers_clean["age"] > 100)
).sum()

print(
    "Invalid ages remaining:",
    remaining_invalid_age
)

print(
    "Null emails:",
    customers_clean["email"].isna().sum()
)

print(
    "Null phone numbers:",
    customers_clean["phone_number"].isna().sum()
)

print(
    "Null PIN codes:",
    customers_clean["pin_code"].isna().sum()
)

print(
    "Null churn dates:",
    customers_clean["churn_date"].isna().sum()
)

print("\nCleaning complete.")
```

# Step 7D — Billing and Payments Cleaning

The billing and payment tables were investigated for duplicate records,
invalid monetary values, incorrect GST, future dates, and broken foreign keys.

Cleaning principles:

- Remove only exact duplicate transaction rows.
- Correct calculable GST values using the expected 18% GST rate.
- Convert invalid negative invoice amounts to null.
- Convert future billing dates to null.
- Preserve payment records with broken invoice references, but set the
  invalid invoice ID to null.
- The original CSV files remain unchanged.
[
          Issue                  	Action
344  - duplicate invoices	 -     Remove exact duplicate rows,
229  - negative amounts       -  	  Convert invalid negative invoice amount fields to NaN rather than deleting invoice,
1,150 -incorrect GST	      -        Recalculate GST from base amount,
176 -  future billing dates	   -   Convert future dates to NaT,
276-duplicate payments	   -   Remove exact duplicate rows,
184 -  broken invoice references	-  Keep payment record but set invalid invoice_id to NaN]

Important: We are preserving the transaction rows wherever possible. We don't want to lose legitimate customer/payment history just because one field is bad.

```python
# ============================================================
# STEP 7D — BILLING CLEANING
# ============================================================
import numpy as np

billing_clean = tables["billing"].copy()

print("BILLING CLEANING")
print("=" * 80)

# ------------------------------------------------------------
# 1. Remove exact duplicate billing rows
# ------------------------------------------------------------

before = len(billing_clean)

billing_clean = billing_clean.drop_duplicates()

removed_duplicates = before - len(billing_clean)

print("Exact duplicate billing rows removed:", removed_duplicates)


# ------------------------------------------------------------
# 2. Convert monetary columns to numeric
# ------------------------------------------------------------

money_columns = [
    "base_amount_inr",
    "gst_amount_inr",
    "total_amount_inr"
]

for col in money_columns:
    billing_clean[col] = pd.to_numeric(
        billing_clean[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# 3. Negative invoice amounts → null
# ------------------------------------------------------------

negative_mask = (
    billing_clean["total_amount_inr"] < 0
)

negative_count = negative_mask.sum()

billing_clean.loc[
    negative_mask,
    [
        "base_amount_inr",
        "gst_amount_inr",
        "total_amount_inr"
    ]
] = np.nan

print("Negative invoice records corrected:", negative_count)


# ------------------------------------------------------------
# 4. Correct GST
# ------------------------------------------------------------

billing_clean["gst_amount_inr"] = (
    billing_clean["base_amount_inr"] * 0.18
).round(2)

# Recalculate total amount
billing_clean["total_amount_inr"] = (
    billing_clean["base_amount_inr"]
    + billing_clean["gst_amount_inr"]
).round(2)

print("GST values recalculated using 18% GST.")


# ------------------------------------------------------------
# 5. Parse billing dates
# ------------------------------------------------------------

billing_clean["billing_date"] = pd.to_datetime(
    billing_clean["billing_date"],
    format="mixed",
    errors="coerce"
)

billing_clean["due_date"] = pd.to_datetime(
    billing_clean["due_date"],
    format="mixed",
    errors="coerce"
)


# ------------------------------------------------------------
# 6. Future billing dates → null
# ------------------------------------------------------------

reference_date = pd.Timestamp("2026-08-24")

future_mask = (
    billing_clean["billing_date"] > reference_date
)

future_count = future_mask.sum()

billing_clean.loc[
    future_mask,
    "billing_date"
] = pd.NaT

print("Future billing dates corrected:", future_count)


print("\nFinal billing rows:", len(billing_clean))
print("Unique invoice IDs:", billing_clean["invoice_id"].nunique())

```

## Usage Data Quality Decision

The usage_data table contains 57,000 records with no missing values or duplicate usage IDs.

An IQR-based analysis identified 4,480 statistical outliers in `data_gb_used`. However, statistical outliers are not automatically data errors and may represent genuinely high-usage customers.

Further investigation identified 166 records with data usage above 500 GB, including several extremely large values. The project's `data_quality_issue_log` identifies 171 affected rows as usage-data outliers.

## Step 7H — Customer Location Cleaning

The investigation found that all customer location IDs are valid and internally consistent. City-to-state and city-tier relationships also contain no mismatches.

A total of **941 customer city names** were found to differ from the authoritative `cities` reference table, primarily due to inconsistent capitalization and spelling variations. These city names will be standardized using the valid `city_id` as the lookup key.

No customer records will be deleted because of location inconsistencies.


```python
# ============================================================
# STEP 7H — CUSTOMER LOCATION CLEANING
# ============================================================

customers_location_clean = tables["customers"].copy()

print("CUSTOMER LOCATION CLEANING")
print("=" * 80)


# ------------------------------------------------------------
# 1. Standardize city names using city_id
# ------------------------------------------------------------

city_name_map = (
    tables["cities"]
    .set_index("city_id")["city_name"]
    .astype(str)
    .to_dict()
)

expected_city_names = (
    customers_location_clean["city_id"]
    .astype(str)
    .map(city_name_map)
)

city_name_mismatch = (
    customers_location_clean["city_name"].astype(str)
    != expected_city_names
)

city_name_correction_count = city_name_mismatch.sum()

customers_location_clean.loc[
    city_name_mismatch,
    "city_name"
] = expected_city_names[city_name_mismatch]

print(
    "City names standardized:",
    city_name_correction_count
)


# ------------------------------------------------------------
# 2. Final validation
# ------------------------------------------------------------

print("\nFINAL CUSTOMER LOCATION VALIDATION")
print("=" * 80)

print(
    "Rows:",
    len(customers_location_clean)
)

print(
    "Unique customer IDs:",
    customers_location_clean["customer_id"].nunique()
)

print(
    "City name mismatches remaining:",
    (
        customers_location_clean["city_name"].astype(str)
        != customers_location_clean["city_id"]
        .astype(str)
        .map(city_name_map)
    ).sum()
)

print(
    "Null city IDs:",
    customers_location_clean["city_id"].isna().sum()
)

print(
    "Null state IDs:",
    customers_location_clean["state_id"].isna().sum()
)

print(
    "Null region IDs:",
    customers_location_clean["region_id"].isna().sum()
)

print(
    "Null city tiers:",
    customers_location_clean["city_tier"].isna().sum()
)
```

```python
# ============================================================
# APPLY LOCATION CLEANING TO FINAL CUSTOMER DATA
# ============================================================

customers_clean["city_name"] = (
    customers_clean["city_id"]
    .astype(str)
    .map(city_name_map)
)

print("Location standardization applied to customers_clean.")
print("Rows:", len(customers_clean))
print("Unique customer IDs:", customers_clean["customer_id"].nunique())
```

```python
# Final city-name validation

city_name_check = (
    customers_clean["city_name"].astype(str)
    == customers_clean["city_id"]
    .astype(str)
    .map(city_name_map)
)

print("City name mismatches remaining:", (~city_name_check).sum())
```

```python
# ============================================================
# SAVE UPDATED CUSTOMER CLEAN DATA
# ============================================================

customers_clean.to_csv(
    "cleaned_data/customers_clean.csv",
    index=False
)

print("Updated customers_clean.csv saved successfully.")
print("File: cleaned_data/customers_clean.csv")
print("Rows:", len(customers_clean))
print("Columns:", len(customers_clean.columns))
```
